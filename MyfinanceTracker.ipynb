```python

import mysql.connector
import tkinter as tk
from tkinter import messagebox

#connecting to the database.
db=mysql.connector.connect(
        host='localhost',
        user='root',
        password='Esibondi@123',
        database='myfinance_tracker'
)



cursor=db.cursor()

# Functions 
def add_income():
    try:
        date= date_entry.get()
        amount= float(amount_entry.get())
        source= source_entry.get()
        query= "INSERT INTO income (date, amount,source) VALUES (%s,%s,%s)"
        cursor.execute(query, (date, amount,source))
        db.commit()
        messagebox.showinfo("Success"," Income added!")
        clear_fields()
    except Exception as e:
        messagebox.showinfo("Error",str(e))
        

def add_expenses():
    try:
        date= date_expend_entry.get()
        amount= float(amount_expend_entry.get())
        category= category_entry.get()
        
        query= "INSERT INTO expenses(date,amount,category,income_id) VALUES( %s,%s,%s)"
        cursor.execute(query,(date,amount,category))
        db.commit()
        messagebox.showinfo("Success","Expense added!")
        clear_exp_fields()
    except Exception as e:
        messagebox.showinfo("Error",str(e))


def view_balance():
    try:
        cursor.execute("SELECT SUM(amount) FROM income")
        total_income = cursor.fetchone()[0] or 0
        cursor.execute = ("SELECT SUM(amount) FROM expenses")
        total_expenses = cursor.fetchone()[0] or 0
        balance = total_income - total_expenses
        messagebox.showinfo("Balance",f"Balance : {balance:.2f}\nIncome:{total_income:.2f}\nExpenses:{total_expenses:.2f}")
    except Exception as e:
        messagebox.showerror("Error",str(e))


def clear_fields():
    date_entry.delete(0,tk.END)
    amount_entry.delete(0,tk.END)
    source_entry.delete(0,tk.END)

def clear_exp_fields():
    date_expend_entry.delete(0,tk.END)
    amount_expend_entry.delete(0,tk.END)
    category_entry.delete(0,tk.END)


#Graphical User Interface.
root = tk.Tk()
root.title("Finance Tracker")


#Income Frame.
income_frame= tk.Frame(root)
income_frame.pack(padx=10,pady=10)
tk.Label(income_frame,text='Add Income').grid(row=0,column=0,columnspan=2)
tk.Label(income_frame,text='Date (YYYY-MM-DD):').grid(row=1,column=0)
date_entry= tk.Entry(income_frame)
date_entry.grid(row=1,column=1)
tk.Label(income_frame,text='Amount').grid(row=2,column=0)
amount_entry=tk.Entry(income_frame)
amount_entry.grid(row=2,column=1)
tk.Label(income_frame,text='Source:').grid(row=3,column=0)
source_entry=tk.Entry(income_frame)
source_entry.grid(row=3,column=1)
tk.Button(income_frame,text='Add income', command=add_income).grid(row=4,column =0,columnspan=2)



#Expenses Frame
expenses_frame=tk.Frame(root)
expenses_frame.pack(padx=10, pady=10)
tk.Label(expenses_frame,text='Add Expenses').grid(row=0,column=0,columnspan=2)
tk.Label(expenses_frame,text='Date (YYYY-MM-DD):').grid(row=1,column=0)
date_expend_entry=tk.Entry(expenses_frame)
date_expend_entry.grid(row=1,column=1)
tk.Label(expenses_frame,text='Amount:').grid(row=2,column=0)
amount_expend_entry=tk.Entry(expenses_frame)
amount_expend_entry.grid(row=2,column=1)
tk.Label(expenses_frame,text='Category:').grid(row=3,column=0)
category_entry=tk.Entry(expenses_frame)
category_entry.grid(row=3,column=1)
tk.Button(expenses_frame,text='Add Expenses',command=add_expenses).grid(row=4,column=0,columnspan=2)

#Balance Button
tk.Button(root,text='View Balance', command=view_balance).pack(pady=10)
root.mainloop()
db.close()
```
